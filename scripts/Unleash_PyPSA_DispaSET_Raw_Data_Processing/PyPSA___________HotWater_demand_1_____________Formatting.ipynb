    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5; border-top: 1px solid rgba(135, 206, 235, 0.3); padding-top: 15px;">
        This routine serves as the entry point for the demand preprocessing engine. It systematically maps raw input data to the required simulation targets through the following mechanisms:
        <br>
        • <b>Intelligent Year Matching:</b> The algorithm performs a proximity search to identify the most relevant historical dataset (the "closest year") for a given target year, ensuring that even if specific annual files are missing, the model maintains high-fidelity coverage.
        <br>
        • <b>Path Construction:</b> It dynamically generates hierarchical file paths for each zone, ensuring that the processed output structure remains consistent and ready for ingestion by the downstream PyPSA simulation environment.
        <br>

        To bridge the gap between raw data and simulation requirements, this procedure reconstructs the DHW demand series into a structured 168-hour weekly cycle. The methodology operates in three distinct phases:
        <br>
        • <b>Baseline Identification:</b> The algorithm aggregates the hourly mean of observed weekday data, establishing a stable baseline representative of typical household water usage patterns.
        <br>
        • <b>Temporal Expansion:</b> A 168-hour scaffold (7 days × 24 hours) is generated, ensuring the pipeline meets the temporal resolution requirements of the PyPSA simulation environment.
        <br>
        • <b>Dynamic Scaling:</b> Utilizing bibliographical scaling factors, the demand profile is adjusted for weekend occupancy (Saturday/Sunday). This ensures that while the weekday baseline is maintained, the peak usage shifts and intensity changes reflect realistic residential water-heating trends.
        <br><br>
                <div style="text-align: justify; color: #87CEEB; font-size: 13px; line-height: 1.5;">
        By synthesizing data through this approach, the pipeline eliminates anomalies inherent in short-term datasets and enforces a scientifically grounded demand structure across all geographical zones.
                </div>
    </div>
</div>

```python
### Define function
def expand_and_scale_profile(vpp_dict):
    """
    1. Expands 5-day data to 7-day (168h) by aggregating daily signatures.
    2. Applies literature-based scaling factors to weekend days.
    """
    factors_df = vpp_dict.get('scaling_factors')
    
    ### ### Iterate through zones in the dictionary
    for key, content in vpp_dict.items():
        if key == 'scaling_factors' or not isinstance(content, dict):
            continue    
        for zone, data in content.items():
            df = data['dataframe'].copy()
            
            ### ### ### Identify columns
            date_col = next(col for col in df.columns if pd.to_datetime(df[col], errors='coerce').notnull().all())
            val_col = next(col for col in df.columns if col != date_col)
            df[date_col] = pd.to_datetime(df[date_col])
            
            ### ### ### Create hourly average profile for Weekdays (0-4)
            weekday_data = df[df[date_col].dt.dayofweek < 5]
            avg_weekday_profile = weekday_data.groupby(weekday_data[date_col].dt.hour)[val_col].mean()
            
            ### ### ### Build the full 168-hour scaffold
            full_week_list = []
            start_date = df[date_col].min().replace(hour=0, minute=0, second=0)
            for day in range(7):
                
                ### ### ### ### Retrieve scaling factor for this specific day_num
                factor = factors_df.loc[factors_df['DayNum'] == day, 'ScalingFactor']
                factor = factor.iloc[0] if not factor.empty else 1.00
                
                ### ### ### ### Apply scaling to the weekday baseline
                scaled_profile = avg_weekday_profile * factor
                
                ### ### ### ### Append to the scaffold
                for hour, value in scaled_profile.items():
                    full_week_list.append({
                        date_col: start_date + pd.Timedelta(days=day, hours=int(hour)),
                        val_col: value
                    })
            
            ### ### ### Update dictionary
            vpp_dict[key][zone]['dataframe'] = pd.DataFrame(full_week_list)
            print(f"✅ Zone {zone}: Successfully expanded to 168 hours and scaled.")
            
    return vpp_dict

### Run the update
vpp_demand_dict = expand_and_scale_profile(vpp_demand_dict)

### Done
print(vpp_demand_dict)
```

```python
import pandas as pd
import numpy as np
import os
from pathlib import Path

# =============================================================================
# USER SETTINGS
# =============================================================================
'''SCENARIO = "Reference_Scenario"
YEAR = "2030"
COUNTRY = "BE"
RESOLUTION = "1h"
root_dir = "/home/ray/Dispa-SET_Unleash/Database_PyPSA"

# Zones or country regions (replace with your list)
zone_names = [COUNTRY]  # or ['BE', 'NL', ...]'''

# Output folder for VPP scenario
output_base_vpp = f"{root_dir}/{SCENARIO}________VPP"

# Heat parameters
COP_SPACE_HEAT = 3.0
SUMMER_MONTHS = [6, 7, 8]
SMOOTHING_HOURS = 24

# =============================================================================
# HELPER FUNCTIONS
# =============================================================================

def load_timeseries(csv_path):
    """Load CSV timeseries robustly"""
    df = pd.read_csv(csv_path, header=None, names=["timestamp", "value"], index_col=0, parse_dates=True)
    return df["value"]

def split_heat_demand_seasonal(heat_series, summer_months=SUMMER_MONTHS, smoothing_window=SMOOTHING_HOURS):
    """
    Split total heat into DHW and space heating (SH)
    """
    heat = heat_series.copy()
    heat.index = pd.to_datetime(heat.index)

    # Summer baseline = DHW
    summer_mask = heat.index.month.isin(summer_months)
    dhw_baseline = heat[summer_mask].mean()
    dhw = pd.Series(dhw_baseline, index=heat.index)

    # Daily profile
    hourly_profile = heat.groupby(heat.index.hour).mean() / heat.groupby(heat.index.hour).mean().mean()
    dhw *= heat.index.hour.map(hourly_profile)

    # Smooth DHW
    dhw = dhw.rolling(smoothing_window, center=True, min_periods=1).mean()

    # Normalize energy after smoothing
    target_energy = dhw_baseline * len(dhw)
    dhw *= target_energy / dhw.sum()

    # Space heating = residual
    sh = heat - dhw
    sh[sh < 0] = 0

    # Final correction
    dhw = heat - sh

    return dhw, sh

# =============================================================================
# PATHS
# =============================================================================
heat_csv_path = f"{root_dir}/{SCENARIO}/Heat_demand/{COUNTRY}/{RESOLUTION}/{YEAR}.csv"
electric_csv_path = f"{root_dir}/{SCENARIO}/Load_RealTime/{COUNTRY}/{RESOLUTION}/{YEAR}.csv"

output_hotwater_path = f"{output_base_vpp}/VPP_data/VPP_demand/{COUNTRY}/{RESOLUTION}/{YEAR}_hotwater_demand.csv"
output_spaceheating_path = f"{output_base_vpp}/VPP_data/VPP_demand/{COUNTRY}/{RESOLUTION}/{YEAR}_spaceheating_demand.csv"
output_electric_path = f"{output_base_vpp}/Load_RealTime/{COUNTRY}/{RESOLUTION}/{YEAR}.csv"

os.makedirs(os.path.dirname(output_hotwater_path), exist_ok=True)
os.makedirs(os.path.dirname(output_spaceheating_path), exist_ok=True)
os.makedirs(os.path.dirname(output_electric_path), exist_ok=True)

# =============================================================================
# LOAD DATA
# =============================================================================
Q_heat_total = load_timeseries(heat_csv_path)
P_electric_collapsed = load_timeseries(electric_csv_path)

# Align indexes
Q_heat_total, P_electric_collapsed = Q_heat_total.align(P_electric_collapsed, join="inner")

# =============================================================================
# SPLIT HEAT DEMAND
# =============================================================================
Q_DHW, Q_SH = split_heat_demand_seasonal(Q_heat_total)

# Convert DHW to electricity
P_DHW_electric = Q_DHW / COP_SPACE_HEAT

# Remove DHW electricity from collapsed power
P_electric_final = P_electric_collapsed - P_DHW_electric

# =============================================================================
# SAVE RESULTS
# =============================================================================
# Save DHW (for VPP)
Q_DHW.to_csv(output_hotwater_path, header=["hot_water_demand"])

# Save space heating (for reference, still inside power demand)
Q_SH.to_csv(output_spaceheating_path, header=["space_heating_demand"])

# Save final collapsed power (without DHW)
P_electric_final.to_csv(output_electric_path, header=False)

# =============================================================================
# VALIDATION CHECKS
# =============================================================================
print("✅ Heat split and power demand updated successfully")
print(f"Saved hot water demand → {output_hotwater_path}")
print(f"Saved space heating demand → {output_spaceheating_path}")
print(f"Saved final collapsed power (without DHW) → {output_electric_path}")

print(f"Energy balance check: {np.allclose(Q_DHW + Q_SH, Q_heat_total)}")
print(f"Power before split: {P_electric_collapsed.sum():.2f}")
print(f"Power after removing DHW: {P_electric_final.sum():.2f}")
print(f"Power difference (DHW removed): {P_electric_collapsed.sum() - P_electric_final.sum():.2f}")
```

```python
import pandas as pd
import numpy as np
import os

# =============================================================================
# USER SETTINGS
# =============================================================================
'''SCENARIO   = "Reference_Scenario"  # e.g., "Sufficiency_Scenario"
YEAR       = "2030"
COUNTRY    = "BE"
RESOLUTION = "1h"
root_dir   = "/home/ray/Dispa-SET_Unleash/Database_PyPSA"'''

# =============================================================================
# PATHS
# =============================================================================

# Input DHW (already split from total heat)
source_dhw_path = f"{root_dir}/{SCENARIO}________VPP/VPP_data/VPP_demand/{COUNTRY}/{RESOLUTION}/{YEAR}_hotwater_demand.csv"

# Reference profile for reshaping DHW
reference_profile_path = "/home/ray/Dispa-SET_Unleash/Database/Boundary_Sector/BoundarySectorDemand/BE/2023.csv"

# Output reshaped DHW (VPP-ready)
output_reshaped_path = f"{root_dir}/{SCENARIO}________VPP/VPP_data/VPP_demand/{COUNTRY}/{RESOLUTION}/{YEAR}.csv"

# Ensure output directories exist
os.makedirs(os.path.dirname(output_reshaped_path), exist_ok=True)

# =============================================================================
# LOAD DHW DATA
# =============================================================================
dhw_df = pd.read_csv(
    source_dhw_path,
    index_col=0,
    parse_dates=True
)

# Extract series
dhw_series = dhw_df.iloc[:, 0]

# Remove timezone if present
dhw_series.index = dhw_series.index.tz_localize(None)

print("Loaded DHW series")
print(f"Total DHW energy: {dhw_series.sum():.2f}")

# =============================================================================
# LOAD REFERENCE PROFILE
# =============================================================================
ref_df = pd.read_csv(
    reference_profile_path,
    index_col=0,
    parse_dates=True
)

ref_series = ref_df.iloc[:, 0]
print("Loaded reference profile")

# =============================================================================
# ALIGN LENGTH WITH DHW SERIES
# =============================================================================
if len(ref_series) != len(dhw_series):
    print("⚠ Length mismatch detected — aligning reference to DHW length")
    if len(ref_series) > len(dhw_series):
        ref_series = ref_series.iloc[:len(dhw_series)]
    else:
        repeats = int(np.ceil(len(dhw_series) / len(ref_series)))
        ref_series = pd.concat([ref_series] * repeats)
        ref_series = ref_series.iloc[:len(dhw_series)]

# Match index
ref_series.index = dhw_series.index

# =============================================================================
# CLEAN & NORMALIZE REFERENCE PROFILE
# =============================================================================
ref_series[ref_series < 0] = 0

if ref_series.sum() == 0:
    raise ValueError("Reference profile has zero total energy.")

ref_norm = ref_series / ref_series.sum()

# =============================================================================
# SCALE REFERENCE TO MATCH DHW ENERGY
# =============================================================================
total_dhw_energy = dhw_series.sum()
dhw_reshaped = ref_norm * total_dhw_energy

# =============================================================================
# SAVE RESHAPED DHW
# =============================================================================
dhw_reshaped.to_csv(
    output_reshaped_path,
    header=["S_1"]
)

# =============================================================================
# VALIDATION CHECKS
# =============================================================================
print("\n✅ DHW reshaping completed")
print(f"Saved → {output_reshaped_path}")
print(f"Original DHW energy: {total_dhw_energy:.2f}")
print(f"Reshaped DHW energy: {dhw_reshaped.sum():.2f}")
print(f"Balanced: {np.isclose(total_dhw_energy, dhw_reshaped.sum())}")
```
