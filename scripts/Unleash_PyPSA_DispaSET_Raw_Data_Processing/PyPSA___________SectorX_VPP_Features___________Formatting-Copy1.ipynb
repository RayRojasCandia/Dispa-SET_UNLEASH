```python
'''# 1. *********************************************************************** Define efficiency / conversion coefficients
heat_coeff = (
    0.5 / 3.0 +   # Heat pumps (COP 3)
    0.2 / 1.0 +   # Resistive heaters
    0.3 * 0.4     # CHP electrical ratio
)

fuel_eff = {
    "oil": 0.35,
    "biomass": 0.30,
    "gas": 0.50,
    "coal": 0.38
}

molecule_eff = {
    "h2": 0.70,
    "methanol": 0.60,
    "nh3": 0.55
}

# 2. ******************************************************************* Helper function to load sector CSV as pd.Series
def load_sector_timeseries(sector, zone):
    try:
        csv_path = collapsed_demand_data_paths_dict[sector][zone]["file_path"]
        df = pd.read_csv(csv_path, index_col=0, parse_dates=True, header=None, names=["timestamp", "demand"])
        return df["demand"]
    except Exception as e:
        print(f"⚠️ Could not load {sector} for {zone}: {e}")
        return pd.Series(dtype=float)

# 3. Loop over zones and collapse Power Load
for zone in zone_names:
    print(f"\nProcessing zone: {zone}")

# 3.1 ***************************************************************************************** Power sector direct load
    power_loads = load_sector_timeseries("Load_RealTime", zone)

# 3.2. **************************************************************************** Heat sector converted to electricity
    heat_loads = load_sector_timeseries("Heat_demand", zone)
    P_heat = heat_loads * heat_coeff if not heat_loads.empty else 0

# 3.3. **************************************************************************************************** Fuel sectors
    oil_loads = load_sector_timeseries("Oil_demand", zone)
    P_oil = oil_loads * fuel_eff["oil"] if not oil_loads.empty else 0

    biomass_loads = load_sector_timeseries("Biomass_demand", zone)
    P_biomass = biomass_loads * fuel_eff["biomass"] if not biomass_loads.empty else 0

    gas_loads = load_sector_timeseries("Gas_Biogas_demand", zone)
    P_gas = gas_loads * fuel_eff["gas"] if not gas_loads.empty else 0

    coal_loads = load_sector_timeseries("Coal_demand", zone)
    P_coal = coal_loads * fuel_eff["coal"] if not coal_loads.empty else 0

# 3.4. ************************************************************************************************ Molecule sectors
    h2_loads = load_sector_timeseries("H2_demand", zone)
    P_h2 = h2_loads / molecule_eff["h2"] if not h2_loads.empty else 0

    methanol_loads = load_sector_timeseries("Methanol_demand", zone)
    P_methanol = methanol_loads / molecule_eff["methanol"] if not methanol_loads.empty else 0

    nh3_loads = load_sector_timeseries("Ammonia_demand", zone)
    P_nh3 = nh3_loads / molecule_eff["nh3"] if not nh3_loads.empty else 0

# 3.5 ******************************************************************************************** Sum all contributions
    P_equivalent = (
        power_loads.fillna(0) +
        P_heat.fillna(0) +
        P_oil.fillna(0) +
        P_biomass.fillna(0) +
        P_gas.fillna(0) +
        P_coal.fillna(0) +
        P_h2.fillna(0) +
        P_methanol.fillna(0) +
        P_nh3.fillna(0)
    )

# 3.6. ********************************************************************** Save to CSV (overwrite original year file)
    output_df = pd.DataFrame({
        "collapsed_power_load": P_equivalent
    })
    csv_path = collapsed_demand_data_paths_dict["Load_RealTime"][zone]["file_path"]
    output_df.to_csv(csv_path, header=False)
    print(f"✅ Collapsed power load saved for {zone} → {csv_path}")

print("\nAll zones processed.")'''
```

Copying the Sector X reference file to the corresponding VPP negaWatt scenario folders

```python
import shutil
import os

# 1. Check if the source file actually exists
if os.path.exists(reference_file_path):
    
    # 2. Extract the directory from the output path
    output_dir = os.path.dirname(output_file_path)
    
    # 3. Create the destination directory if it doesn't exist
    if output_dir and not os.path.exists(output_dir):
        os.makedirs(output_dir)
        
    # 4. Perform the copy
    # copy2 preserves metadata like timestamps
    shutil.copy2(reference_file_path, output_file_path)
    print(f"File copied successfully to {output_file_path}")

else:
    print(f"Error: Source file not found at {reference_file_path}")
```

```python
import pandas as pd


# Read the CSV files
vpp_df = pd.read_csv(vpp_unit_file_path)
output_df = pd.read_csv(output_file_path)

# Search for the value in the 'Unit' column
search_value = f"VPP_{zone_names}"
matching_row_vpp = vpp_df[vpp_df['Unit'] == search_value]

if not matching_row_vpp.empty:
    # Get the corresponding 'Sector' value from the output file
    sector_value = matching_row_vpp['Sector1'].values[0]
    matching_row_output = output_df[output_df['Sector'] == sector_value]

    if not matching_row_output.empty:
        # Update the 'STOCapacity' and 'STOMaxPower' columns
        output_df.loc[output_df['Sector'] == sector_value, 'STOCapacity'] = matching_row_vpp['STOCapacity'].values[0]
        output_df.loc[output_df['Sector'] == sector_value, 'STOMaxPower'] = matching_row_vpp['STOMaxChargingPower'].values[0]

        # Save the updated DataFrame back to the CSV file
        output_df.to_csv(output_file_path, index=False)
        print("Data updated successfully.")
    else:
        print("No matching 'Sector' found in the output file.")
else:
    print("No matching 'Unit' found in the VPP unit file.")
```
