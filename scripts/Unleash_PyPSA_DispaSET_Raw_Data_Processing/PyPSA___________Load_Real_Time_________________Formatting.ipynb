```python
# 1. ************************************************************************** Extract a 4-digit year from the filename
def extract_year_from_filename(filename):
    match = re.search(r"\b(19|20)\d{2}\b", filename)
    return int(match.group()) if match else None

# 2. ********************************************* Iterate over each demand sector and its associated configuration data
for sector_key, sector_data in demand_sectors_formatted_data_folder_paths_dict.items():

# 2.1 ************************************************************************** Skip sectors without a base folder path
    if "folder_path" not in sector_data:
        continue

    base_folder_path = sector_data["folder_path"]

# 2.2. ******************************************************************************** Container for country-level data
    sector_data["countries"] = {}

    for zone in zone_names:

# 2.2.1. ******************************************************************* Folder where real-time CSV files are stored
        csv_folder_path = os.path.join(
            load_real_time_base_data_folder_path,
            zone,
            data_target_time_step
        )

# 2.2.2. ************************************************************** Folder path to be stored in the output structure
        country_folder_path = os.path.join(
            base_folder_path,
            zone,
            data_target_time_step
        )

        closest_csv = None
        closest_year_diff = float("inf")

# 2.2.3. ***************************************************Search for the CSV whose year is closest to data_target_year
        if os.path.isdir(csv_folder_path):
            for file in os.listdir(csv_folder_path):

                if not file.lower().endswith(".csv"):
                    continue

                year = extract_year_from_filename(file)
                if year is None:
                    continue

# 2.2.3.1. ******************************************************************** Compare year distance to the target year
                target_year = int(data_target_year)
                diff = abs(year - target_year)
                if diff < closest_year_diff:
                    closest_year_diff = diff
                    closest_csv = os.path.join(csv_folder_path, file)

# 2.2.4. ************************************************************************ Load the selected CSV into a DataFrame
        if closest_csv is not None:
            df = pd.read_csv(closest_csv, header=None, names=['timestamp', 'demand'])
            country_demand_df = {data_target_year: df}

        else:

# 2.2.5. ***************************************************************************************** No suitable CSV found
            country_demand_df = {data_target_year: None}

# 2.2.6. ************************************************************************* Store results for the current country
        sector_data["countries"][zone] = {
            "country_folder_path": country_folder_path,
            "country_demand_df": country_demand_df
        }

# 2. ************************************************************************************************************* Done 
print(f"Demand Sectors Formatted Data Folder Paths Dictionary:        {demand_sectors_formatted_data_folder_paths_dict}")
```

5.2. Time Step Correction
</div>
<div style="text-align: justify; margin-left: 0.0em; font-weight: unbold; font-size: 14px; font-family: TimesNewRoman; color:skyblue">
Verifying and correcting first column timestamps in the dataframes to match target year and selected time step.
<hr style="border: 1px solid skyblue;">

```python
# 1. **************************************** Find the first column that can be converted to datetime without NaN values
def find_timestamp_column(df: pd.DataFrame) -> str:
    for col in df.select_dtypes(include=["object", "datetime"]):
        if pd.to_datetime(df[col], errors="coerce").notna().all():
            return col
    raise ValueError("No timestamp column found")

# 2. **************************************************************************************** Update the time step date
def update_dataframe_year(df: pd.DataFrame, target_year: int) -> pd.DataFrame:
    df = df.copy()
    ts_col = find_timestamp_column(df)
    df[ts_col] = pd.to_datetime(df[ts_col], utc=True)
    
    source_year = df[ts_col].dt.year.iloc[0]
    
# 2.1. ****************************************************************************************** Adjust for leap years
    if calendar.isleap(source_year) and not calendar.isleap(target_year):
        
# 2.1.1. ************************************************************************************************ Remove Feb 29
        df = df[~((df[ts_col].dt.month == 2) & (df[ts_col].dt.day == 29))]

    elif not calendar.isleap(source_year) and calendar.isleap(target_year):

# 2.1.2. ******************************************************************************* Duplicate Feb 28 for leap year
        feb28 = df[(df[ts_col].dt.month == 2) & (df[ts_col].dt.day == 28)]
        df = pd.concat([df, feb28], ignore_index=True)

# 2.2. ************************************************************************************ Vectorized year replacement
    df[ts_col] = df[ts_col] + pd.offsets.DateOffset(years=(target_year - source_year))
    
    return df

# 3. ******************************************************* Recursively update all DataFrames in the nested dictionary
def update_all_dfs(nested_dict, target_year):
    for sector in nested_dict.values():
        countries = sector.get("countries", {})
        for country_code, country_data in countries.items():
            for sub_dict in country_data.values():
                if isinstance(sub_dict, dict):
                    for key, df in sub_dict.items():
                        if isinstance(df, pd.DataFrame):
                            sub_dict[key] = update_dataframe_year(df, target_year)

update_all_dfs(demand_sectors_formatted_data_folder_paths_dict, int(data_target_year))

# 2. ************************************************************************************************************* Done 
print(f"Demand Sectors Formatted Data Folder Paths Dictionary:        {demand_sectors_formatted_data_folder_paths_dict}")
```

```python
# 1. ************************************************************* Identify the timestamp column to keep it at the front
time_col = load_pypsa_raw_data_df.columns[0]

# 2. ************************************************************************ Define the "Cleaning" function for headers
def clean_header_logic(col_name):
    if col_name == time_col:
        return time_col    
    match = re.search(r"^\s*([A-Z]{2})\d*\s+\d+\s+(.*)", col_name)
    if match:
        country_code = match.group(1)
        sector_description = match.group(2)

# 2.2. **************************************************** *Combine columns into the "National" requested header format
        return f"{country_code} {sector_description}"
    
# 2.3. ******************************************************************* Fallback if the pattern doesn't match exactly
    return col_name.strip()

# 3. ********************************************************************** Apply the GroupBy along the columns (axis=1)
load_pypsa_aggregated_df = load_pypsa_raw_data_df.groupby(clean_header_logic, axis=1).sum()

# 4. *************************************************************** Reorder to ensure the timestamp is the first column
cols = list(load_pypsa_aggregated_df.columns)
if time_col in cols:

# 4.1. **************************************************************************************** Move time_col to index 0
    cols.insert(0, cols.pop(cols.index(time_col)))
    load_pypsa_aggregated_df = load_pypsa_aggregated_df[cols]

# 5. ************************************************************************************************************** Done 
print("Headers after sum:")
for col in load_pypsa_aggregated_df.columns[:5]:
    print(f" - '{col}'")
print(load_pypsa_aggregated_df)
```
