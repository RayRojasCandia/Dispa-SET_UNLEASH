<div style="background-color: black;">
<hr style="border: 2px solid skyblue;">
<div style="text-align: justify; margin-left: 3.0em; font-weight: bold; font-size: 17px; font-family: TimesNewRoman; color:skyblue">
5.2. Empty Zone Data Frames
</div>
<div style="text-align: justify; margin-left: 0.0em; font-weight: unbold; font-size: 14px; font-family: TimesNewRoman; color:skyblue">
Creating empty dataframes—column headers only—for the selected zones, corresponding to the target year
<hr style="border: 1px solid skyblue;">
</div>

```python
# 1. ------------------------------------------------------------------------------------ convert data year to integer 
data_target_year = int(data_target_year)  

# 2. -------------------------------------------------------------------------- Dictionary to store created DataFrames 
scaled_inflows_dfs_dict = {}

for zone in zone_names:
    zone_folder = os.path.join(scaled_inflows_base_data_folder_path, zone)
    if not os.path.isdir(zone_folder):
        continue  # skip if zone folder doesn't exist

    # 2.1. ---------------------------------------------------------- Go into subfolder matching data_target_time_step 
    timestep_folder = os.path.join(zone_folder, str(data_target_time_step))
    if not os.path.isdir(timestep_folder):
        continue  # skip if subfolder doesn't exist

    # 2.2. ----------------------------------------------------------------- List all csv files in the timestep folder 
    csv_files = [f for f in os.listdir(timestep_folder) if f.endswith(".csv")]

    # 2.3 ---------------------------------------------------------- Extract years from filenames (must be digits only) 
    available_years = []
    for f in csv_files:
        name, ext = os.path.splitext(f)
        if name.isdigit():  # e.g., "2004"
            available_years.append(int(name))

    if not available_years:
        continue  # skip if no year-based CSVs exist

    # 2.3 --------------------------------------------------------------------------------- Find closest year to target 
    closest_year = min(available_years, key=lambda y: abs(y - data_target_year))

    # 2.4 ------------------------------------------------------------------------------------- Path to the chosen file 
    chosen_file = os.path.join(timestep_folder, f"{closest_year}.csv")

    # 2.5 ---------------------------------------------------------------------------------- Read only the first column 
    first_col_name = pd.read_csv(chosen_file, nrows=0).columns[0]  # get first column name
    first_col_data = pd.read_csv(chosen_file, usecols=[first_col_name])

    # 2.6 -------------------------------------------------------------------------------------------- Read all headers 
    all_headers = pd.read_csv(chosen_file, nrows=0).columns.tolist()

    # 2.7 ----------------------------------------------- Create new DataFrame: first column has data, others are empty 
    df = pd.DataFrame(columns=all_headers)
    df[first_col_name] = first_col_data[first_col_name]

    # 2.8 ------------------------------------------------------- Store in dictionary and optionally as global variable 
    df_name = f"{zone}_{data_target_year}"
    scaled_inflows_dfs_dict[df_name] = df
    globals()[df_name] = df

# 3. ------------------------------------------------------------------------------------------------------------ Done 
    print(f"Zone {zone}: picked {closest_year}.csv for target {data_target_year} and copied first column")

print("Scaled Inflows Data Frames:", list(scaled_inflows_dfs_dict.keys()))
```

```python
# 1. ---------------------------------------------- Define a mapping for time step strings to pandas frequency strings 
time_step_map = {
    
                   "1h"   : "H",
                   "15min": "15T",
                   "30min": "30T"

                }

# 2. -------------------------------------------------------- Get the pandas frequency string for the target time step 
freq = time_step_map[data_target_time_step]

for df_name, df in scaled_inflows_dfs_dict.items():
    if df.empty:
        continue  # skip empty dataframes
    
    # 2.1 ---------------------------------------------------------------------------------- Identify the first column 
    first_col = df.columns[0]
    
    # 2.2 ---------------------------------------------------------- Convert column to datetime with UTC if not already 
    df[first_col] = pd.to_datetime(df[first_col], utc=True, errors='coerce')

    # 2.3 ------------------------------------------------------------------------ Remove any rows that failed to parse 
    df = df.dropna(subset=[first_col])

    # 2.4 ------------------------------------------------------ Create a date range for the correct year and time step 
    start_time = pd.Timestamp(f"{data_target_year}-01-01 00:00:00", tz="UTC")
    end_time = pd.Timestamp(f"{data_target_year}-12-31 23:59:59", tz="UTC")

    correct_index = pd.date_range(start=start_time, end=end_time, freq=freq)

    # 2.5 ------------------------------------------------------ Replace the first column with the corrected date range 
    if len(correct_index) >= len(df):
        df[first_col] = correct_index[:len(df)]
    else:
        # 2.5.1 ----------------------------------- If df has more rows than the date range, extend with repeated values 
        repeats = (len(df) // len(correct_index)) + 1
        df[first_col] = pd.Series(list(correct_index) * repeats)[:len(df)]

    # 2.6 ----------------------------------------------------------------------- Update the DataFrame in the dictionary 
    scaled_inflows_dfs_dict[df_name] = df
    globals()[df_name] = df  # optional if you use global variables

# 3. ------------------------------------------------------------------------------------------------------------ Done 
    print(f"Updated {df_name}: first column aligned with {data_target_year} and {data_target_time_step}")
```

    <ul style="margin-top: 0; padding-left: 1.5em;">
        <li>
        <strong>ΔS:</strong> Change in reservoir storage (energy).
        </li>
        <li>
        <strong>Q<sub>p</sub>:</strong> Pumping Inflow (auto inflow). Water actively moved from the lower to the upper reservoir.
        </li>
        <li>
        <strong>R:</strong> Natural Inflow. Direct rainfall onto the reservoir surface plus runoff from the surrounding catchment area.
        </li>
        <li>
        <strong>Q<sub>g</sub>:</strong> Generation Outflow. Water released through turbines to produce electricity.
        </li>
        <li>
        <strong>E:</strong> Evaporation. Loss of water to the atmosphere.
        </li>
        <li>
        <strong>L:</strong> Seepage/Leakage. Loss of water through the reservoir bed or dam structure.
        </li>
    </ul>
    </div>
    <div style="text-align: justify; margin-left:2em; font-weight: unbold; font-size: 12px; font-family: TimesNewRoman; color:skyblue">
    * Notes: &nbsp;&nbsp; Due to a lack of available data, <strong>E</strong> and <strong>L</strong> losses are considered negligible and are omitted from the calculation.

7. Raw Data Uploading
    </div>
    <div style="text-align: justify; margin-left: 0.0em; font-weight: unbold; font-size: 14px; font-family: TimesNewRoman; color:skyblue">
The Scalled Inflows time series will be constructed from the hourly PyPSA emergy stored values, the energy dispached (outflows) and the energy pumped (autoinglows) for the conventional hydro technologies.
    <br>
 These values will be stored in dictionaries partitioned by country.
    </div>
<hr style="border: 0.5px solid skyblue;">

```python
# 1. ---------------------------------------------- Return the first CSV file in folder containing the year in its name
def find_file_with_year(folder_path, year):
    for f in os.listdir(folder_path):
        if f.endswith('.csv') and str(year) in f:
            return os.path.join(folder_path, f)
    raise FileNotFoundError(f"No file with year {year} in {folder_path}")

# 2. ---------------------------------------------------------- Get all zone variations including numbered and acronyms
def get_zone_variations(zone, zone_names_equivalences_dict):
    variations = [zone] + [f"{zone}{i}" for i in range(10)]
    acronyms = zone_names_equivalences_dict.get(zone, {}).get('Acronym', [])
    if isinstance(acronyms, str):
        acronyms = [acronyms]
    for acr in acronyms:
        variations.append(acr)
        variations.extend([f"{acr}{i}" for i in range(10)])
    return variations
    
# 3. -------------------------------------------------------------------- Find columns that match zone + value pattern
def find_matching_columns(df_columns, zone_variations, values):
    matched_cols = []
    for col in df_columns:
        for zone in zone_variations:
            for val in values:
                pattern = f"{zone} 0 {val}"
                if pattern.lower() in str(col).lower():
                    matched_cols.append((col, val))
                    break
            if matched_cols and matched_cols[-1][0] == col:
                break
    return matched_cols

# 4. -------------------------------------------------------------------------------------------- Load CSV Files Once
energy_file = find_file_with_year(energy_stored_pypsa_raw_data_folder_path, data_target_year)
energy_df = pd.read_csv(energy_file)

hydro_file = find_file_with_year(hydro_dispatch_pypsa_raw_data_folder_path, data_target_year)
hydro_df = pd.read_csv(hydro_file)

# 5. ---------------------------------------------------------------------------------------------- Process Each Zone
for zone in zone_names:
    df_key = f"{zone}_{data_target_year}"
    if df_key not in scaled_inflows_dfs_dict:
        continue
    
    base_df = scaled_inflows_dfs_dict[df_key].copy()
    base_rows = len(base_df)
    zone_variations = get_zone_variations(zone, zone_names_equivalences_dict)
    
    for tech_key, tech_info in tech_equivalences_dict.items():
        if tech_key not in base_df.columns:
            continue
        
        # 5.1. --------------------------------------------------------------------------------------------- Storage
        if tech_info.get('storage') and energy_df is not None:
            storage_values = tech_info['storage'] if isinstance(tech_info['storage'], list) else [tech_info['storage']]
            for orig_col, val in find_matching_columns(energy_df.columns, zone_variations, storage_values):
                if len(energy_df[orig_col]) != base_rows:
                    warnings.warn(f"Row mismatch for {zone}-{tech_key}-storage")
                    continue
                base_df[f"{tech_key}_storage"] = energy_df[orig_col].values
        
        # 5.2. --------------------------------------------------------------------------------------------- Outflow
        if tech_info.get('outflow') and hydro_df is not None:
            outflow_values = tech_info['outflow'] if isinstance(tech_info['outflow'], list) else [tech_info['outflow']]
            for orig_col, val in find_matching_columns(hydro_df.columns, zone_variations, outflow_values):
                if len(hydro_df[orig_col]) != base_rows:
                    warnings.warn(f"Row mismatch for {zone}-{tech_key}-outflow")
                    continue
                base_df[f"{tech_key}_outflow"] = hydro_df[orig_col].values

        # 5.3. ------------------------------------------------------------------------------------------ Autoinflow
        if tech_info.get('autoinflow') and tech_info['autoinflow'] != "-":
            autoinflow_values = tech_info['autoinflow'] if isinstance(tech_info['autoinflow'], list) else [tech_info['autoinflow']]
            year_subfolder = os.path.join(dispatch_pypsa_raw_data_folder_path, str(data_target_year))
        
            for f in os.listdir(year_subfolder):
                if f.endswith('.csv') and any(z.upper() in f.upper() for z in zone_variations):
                    dispatch_df = pd.read_csv(os.path.join(year_subfolder, f))
                    break
            else:
                dispatch_df = pd.DataFrame(index=range(base_rows))
        
            for val in autoinflow_values:
                matched = [c for c in dispatch_df.columns if val.lower() in str(c).lower()]
                if matched:
                    for orig_col in matched:
                        if len(dispatch_df[orig_col]) != base_rows:
                            warnings.warn(f"Row mismatch for {zone}-{tech_key}-autoinflow")
                            continue
        
                        series = dispatch_df[orig_col].values
                        series = np.where(series < 0, -series, 0)
                        base_df[f"{tech_key}_autoinflow"] = series
                        
                else:
                    base_df[f"{tech_key}_autoinflow"] = 0

    scaled_inflows_dfs_dict[df_key] = base_df

# 6. ------------------------------------------------------------------------------------------------------------ Done 
scaled_inflows_dfs_dict
```

    8. Hydro Inflows Calculus
    </div>
    <div style="text-align: justify; margin-left: 0.0em; font-weight: unbold; font-size: 14px; font-family: TimesNewRoman; color:skyblue">
The scaled inflow values will be calculated using the Water Balance Equation previously defined.
    <br>
These values will be stored in a dictionary partitioned by country.

```python
# 1. --------------------------------------------------- Identify existing source columns over each zone-year dataframe
for zone in zone_names:
    df_key = f"{zone}_{data_target_year}"
    if df_key not in scaled_inflows_dfs_dict:
        continue

    df = scaled_inflows_dfs_dict[df_key]

    for tech_key in tech_equivalences_dict.keys():
        storage_col = f"{tech_key}_storage"
        outflow_col = f"{tech_key}_outflow"
        autoinflow_col = f"{tech_key}_autoinflow"

        source_cols = [storage_col, outflow_col, autoinflow_col]
        existing_source_cols = [c for c in source_cols if c in df.columns]

        # 1.1. -------------------------------------------------------------------------------- If nothing exists, skip
        if not existing_source_cols:
            continue

        # 1.2. ------------------------------------------------------------------ If ALL source columns exist → compute
        if len(existing_source_cols) == 3:
            storage = df[storage_col]
            outflow = df[outflow_col]
            autoinflow = df[autoinflow_col]

            storage_prev = storage.shift(1)
            storage_prev.iloc[0] = storage.iloc[-1]

            df[tech_key] = (
                storage
                - storage_prev
                + outflow
                - autoinflow
            )

        # 1.3. -------------------------------------------------------------------- ALWAYS drop existing source columns
        df.drop(columns=existing_source_cols, inplace=True)

    scaled_inflows_dfs_dict[df_key] = df

# 2. ------------------------------------------------------------------------------------------------------------ Done 
scaled_inflows_dfs_dict
```

```python
# 1. -------------------------------------------------- Replace negative values by zero in resultant technology columns
for zone in zone_names:
    df_key = f"{zone}_{data_target_year}"
    if df_key not in scaled_inflows_dfs_dict:
        continue

    df = scaled_inflows_dfs_dict[df_key]

    for tech_key in tech_equivalences_dict.keys():

        # 1.1. -------------------------------------------------------- Apply clipping only if resultant column exists
        if tech_key in df.columns:
            df[tech_key] = df[tech_key].clip(lower=0)

    scaled_inflows_dfs_dict[df_key] = df

# 2. ------------------------------------------------------------------------------------------------------------ Done
scaled_inflows_dfs_dict
```

    9. Scaled Inflows Formating
    </div>
    <div style="text-align: justify; margin-left: 0.0em; font-weight: unbold; font-size: 14px; font-family: TimesNewRoman; color:skyblue">
The hydro input to dispaset is defined as “StorageInflows”.
<br>
It is the normalized values of the inflow with respect to the nominal power of the storage unit—in discharge mode. 
<br>
As an example, if the inflow value at a certain time is 100MWh/h and if the turbining capacity of the hydro plant is 200 MW, the scaled inflow value must be defined as 0.5.
    </div>
<div style="text-align: justify; margin-left: 5em; font-weight: unbold; font-size: 14px; font-family: TimesNewRoman; color:skyblue">
    <ol style="margin-top: 0; padding-left: 1.5em;">
        <li>
            The Dispa-SET documentation.
            <br>
            <a href="https://www.dispaset.eu/en/latest/data.html" style="color:skyblue">https://www.dispaset.eu/en/latest/data.html</a>
        </li>
</div>
<hr style="border: 0.5px solid skyblue;">
</div>

```python
for zone in zone_names:
    # 1. ------------------------------------------------------------------------------ Identify dataframe and CSV file 
    df_key = f"{zone}_{data_target_year}"

    if df_key not in scaled_inflows_dfs_dict:
        continue  # No dataframe for this zone/year

    df = scaled_inflows_dfs_dict[df_key]

    csv_path = os.path.join(
        power_plants_pypsa_formated_data_folder_path,
        zone,
        f"{data_target_year}.csv"
    )

    if not os.path.isfile(csv_path):
        continue  # No corresponding CSV file

    plants_df = pd.read_csv(csv_path)

    # 2. -------------------------------------------------------------------------------- Ensure required columns exist 
    if not {"Technology", "PowerCapacity"}.issubset(plants_df.columns):
        continue

    # 3. ------------------------------------------------------------ Precompute available technologies for fast lookup 
    available_technologies = set(plants_df["Technology"].dropna().unique())

    # 4. -------------------------------------------------------- Iterate over named columns in the reservoir dataframe 
    for col in df.columns:
        
        # 4.1. ------------------------------------------------------------------------- Skip unnamed / invalid columns
        if col is None or (isinstance(col, float) and np.isnan(col)):
            continue

        # 4.2. ----------------------------------------------------------------- Skip columns that are not technologies
        if col not in available_technologies:
            continue

        # 4.3. -------------------------------------------------------------- Sum PowerCapacity for matching Technology 
        matching_rows = plants_df[plants_df["Technology"] == col]

        total_capacity = matching_rows["PowerCapacity"].sum()

        # 4.4. ----------------------------------------------------------------------------- Normalize dataframe column 
        if total_capacity > 0:
            df[col] = df[col] / total_capacity
            df[col] = df[col].clip(upper=1) # Cap values at 1
        else:
            # 4.4.1. ------------------------------------------------ Avoid division by zero; choose preferred behavior
            df[col] = np.nan  # or leave unchanged, or set to 0

# 5. ------------------------------------------------------------------------------------------------------------- Done
scaled_inflows_dfs_dict
```

```python
# 1. ----------------------------------------------------------------- Iterate over each key in scaled_inflows_dfs_dict
for key in scaled_inflows_dfs_dict:
    # 1.1. ------------------------------------------ Extract the zone acronym from the key (e.g., "BE" from "BE_2030")
    zone = key.split('_')[0]

    # 1.2. ---------------------------------------------------------------------- Construct the path to the zone folder
    zone_folder_path = os.path.join(scaled_inflows_pypsa_formated_data_folder_path, zone)

    # 1.3. ---------------------------------------------------------------------------- Check if the zone folder exists
    if not os.path.exists(zone_folder_path):
        print(f"Zone folder does not exist: {zone_folder_path}")
        continue

    # 1.4. -------------------------------------------------- Construct the path to the data_target_time_step subfolder
    time_step_folder_path = os.path.join(zone_folder_path, data_target_time_step)

    # 1.5 ---------------------------------------------- Create the data_target_time_step subfolder if it doesn't exist
    if not os.path.exists(time_step_folder_path):
        os.makedirs(time_step_folder_path)
        print(f"Created subfolder: {time_step_folder_path}")

    # 1.6 ---------------------------------------------------------------------------- Define the path for the CSV file
    csv_file_path = os.path.join(time_step_folder_path, f"{data_target_year}.csv")

    # 1.7 -------------------------------------------------------------------------- Save the DataFrame to the CSV file
    scaled_inflows_dfs_dict[key].to_csv(csv_file_path, index=False)
    print(f"Saved DataFrame to: {csv_file_path}")

# 2. ------------------------------------------------------------------------------------------------------------- Done
print("All DataFrames saved successfully!")
```
