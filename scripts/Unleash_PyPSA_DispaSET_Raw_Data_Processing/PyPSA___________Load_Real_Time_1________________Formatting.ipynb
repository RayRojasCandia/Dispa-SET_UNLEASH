```python
# 1. ******************************************************************************************** Identify the CSV file
csv_file = None
for file in os.listdir(load_real_time_pypsa_raw_data_folder_path):
    if 'link_consumption' in file and str(data_target_year) in file and file.endswith('.csv'):
        csv_file = file
        break

if not csv_file:
    raise FileNotFoundError(f"No matching CSV found in {load_real_time_pypsa_raw_data_folder_path}")

csv_path = os.path.join(load_real_time_pypsa_raw_data_folder_path, csv_file)

# 2. **************************************************************************************** Load and Format Timestamp
df_raw = pd.read_csv(csv_path)
first_col_name = df_raw.columns[0]
df_raw[first_col_name] = pd.to_datetime(df_raw[first_col_name]).dt.strftime('%Y-%m-%d %H:%M:%S')

# 3. ********************************************************************************** Process Zones and Clean Headers
link_consumption_dict = {}

for zone in zone_names:
    
# 3.1. ************************************************************************************* Retrieve potential Acronym
    acronym_list = zone_names_equivalences_dict.get(zone, {}).get("Acronym", [])
    acronym = acronym_list[0].strip() if acronym_list and acronym_list[0].strip() else None
    
# 3.2. ************************************************************************************** Identify matching columns
    matched_cols = [col for col in df_raw.columns if zone in col]
    string_to_erase = zone # Default string to remove
    
# 3.3. *************************************************************** Fallback to Acronym if no columns found for Zone
    if not matched_cols and acronym:
        matched_cols = [col for col in df_raw.columns if acronym in col]
        string_to_erase = acronym # Update string to remove
    
    if matched_cols:
        
# 3.3.1. ***************************************************************************************** Create the DataFrame
        zone_df = df_raw[[first_col_name] + matched_cols].copy()
        
# 3.3.2. ************************************************************************************************ Clean headers
        zone_df.columns = [col.replace(string_to_erase, "").strip("_ ") if col != first_col_name else col for col in zone_df.columns]
        
# 3.3.3. ****************************************************************************************** Store in dictionary
        link_consumption_dict[zone] = zone_df

# 4. ************************************************************************************************************** Done
print(f"Dictionary 'link_consumption_dict' ready. Headers have been cleaned of Zone/Acronym identifiers.")
print(link_consumption_dict)
```

```python
# 1. **************************************** Iterate through sectors and zones to locate and store valid CSV data paths
demand_data_paths_dict = {}

for sector, sector_info in demand_sectors_formatted_data_folder_paths_dict.items():

    sector_folder_path = sector_info["folder_path"]

# 1.1. ***************************************************************************************** Initialize sector level
    demand_data_paths_dict[sector] = {}

    for zone in zone_names:

# 1.1.1. ************************************************************************* Build path: Sector → Zone → Time step
        zone_path = os.path.join(
            sector_folder_path,
            zone,
            data_target_time_step
        )

# 1.1.2. *************************************************************************** Expected CSV file (year = filename)
        csv_file_path = os.path.join(
            zone_path,
            f"{data_target_year}.csv"
        )

# 1.1.3. *********************************************************************************************** Check existence
        if not os.path.exists(csv_file_path):
            print(f"⚠️ Missing file: {csv_file_path}")
            continue

# 1.1.4. **************************************************************************************************** Store info
        demand_data_paths_dict[sector][zone] = {
            "folder_path": zone_path,
            "file_path": csv_file_path
        }

        print(f"Loaded → Sector: {sector} | Zone: {zone}")

# 2. ************************************************************************************************************ Done 
print("\nDictionary successfully built.")
print(demand_data_paths_dict)
```

This cell consolidates the 11 individual demand sectors into a single equivalent power demand profile

```python
import pandas as pd

# *************************
# ELECTRIFICATION ASSUMPTIONS

# Heat electrification
heat_pump_share = 0.8
resistive_share = 0.2
average_cop = 3.0

heat_coeff = (
    heat_pump_share / average_cop +
    resistive_share / 1.0
)

# Molecule production efficiencies
molecule_eff = {
    "h2": 0.70,
    "methanol": 0.60,
    "nh3": 0.55
}

# *************************
# Helper

def load_sector_timeseries(sector, zone):
    try:
        csv_path = collapsed_demand_data_paths_dict[sector][zone]["file_path"]
        df = pd.read_csv(
            csv_path,
            index_col=0,
            parse_dates=True,
            header=None,
            names=["timestamp", "demand"]
        )
        return df["demand"]
    except Exception as e:
        print(f"⚠️ Could not load {sector} for {zone}: {e}")
        return pd.Series(dtype=float)

# *************************
# Main Loop

for zone in zone_names:
    print(f"\nProcessing zone: {zone}")

    # Direct electricity demand
    power_loads = load_sector_timeseries("Power_demand", zone)

    # Heat electrified
    heat_loads = load_sector_timeseries("Heat_demand", zone)
    P_heat = heat_loads * heat_coeff if not heat_loads.empty else 0

    # Molecules (electricity required)
    h2_loads = load_sector_timeseries("H2_demand", zone)
    P_h2 = h2_loads / molecule_eff["h2"] if not h2_loads.empty else 0

    methanol_loads = load_sector_timeseries("Methanol_demand", zone)
    P_methanol = methanol_loads / molecule_eff["methanol"] if not methanol_loads.empty else 0

    nh3_loads = load_sector_timeseries("Ammonia_demand", zone)
    P_nh3 = nh3_loads / molecule_eff["nh3"] if not nh3_loads.empty else 0

    # Collapse ONLY demand-side sectors
    P_equivalent = (
        power_loads.fillna(0) +
        (P_heat.fillna(0) if isinstance(P_heat, pd.Series) else 0) +
        (P_h2.fillna(0) if isinstance(P_h2, pd.Series) else 0) +
        (P_methanol.fillna(0) if isinstance(P_methanol, pd.Series) else 0) +
        (P_nh3.fillna(0) if isinstance(P_nh3, pd.Series) else 0)
    )

    # Save
    output_df = pd.DataFrame({
        "collapsed_power_load": P_equivalent
    })

    csv_path = collapsed_demand_data_paths_dict_1["Load_RealTime"][zone]["file_path"]
    output_df.to_csv(csv_path, header=False)

    print(f"✅ Collapsed power load saved for {zone} → {csv_path}")

print("\nAll zones processed.")
```
