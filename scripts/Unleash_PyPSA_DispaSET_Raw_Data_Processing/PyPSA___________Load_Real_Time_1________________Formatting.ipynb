5.2. Dynamic Load Data Retrieval and Year Alignment

    </div>
    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5; border-top: 1px solid rgba(135, 206, 235, 0.3); padding-top: 15px;">
        This block initializes the <code>load_real_time_dict</code> by identifying and loading the most appropriate historical or forecast load data for each specified zone. The process involves:
        <br>
        • <b>Closest Year Matching:</b> The script scans the target directory for available CSV files and uses a minimization function (<code>min(years, key=lambda x: abs(x - data_target_year_int))</code>) to select the file corresponding to the year closest to your target scenario.
        <br>
        • <b>Automated Data Loading:</b> It reads the identified CSV into a <code>Pandas DataFrame</code>, ensuring the environment is prepared for immediate processing.
        <br>
        • <b>Standardized Schema:</b> The DataFrame columns are programmatically renamed to <code>['snapshot', 'load_real_time']</code>, ensuring that subsequent summation and integration tasks operate on a predictable data structure regardless of the raw CSV's original formatting.
        <br>
        <br>
    <div style="text-align: justify; color: #87CEEB; font-size: 13px; line-height: 1.5;">
        This robust retrieval logic ensures that even if a perfect match for the target year is unavailable, the system intelligently selects the most representative data available, maintaining the continuity of the energy modeling workflow.
    </div>
</div>

```python
### convert data year to integer
data_target_year_int = int(data_target_year)

### Initialize the dictionary to store DataFrames
load_real_time_dict = {}

for zone in zone_names:

    ### ### Construct the path to the zone's time_step folder
    zone_path = os.path.join(load_real_time_base_data_folder_path, zone, data_target_time_step)

    ### ### List all CSV files in the time_step folder
    csv_files = [f for f in os.listdir(zone_path) if f.endswith('.csv')]

    ### ### Extract years from CSV filenames
    years = []
    for csv_file in csv_files:
        year_str = csv_file.split('.')[0]
        if year_str.isdigit():
            years.append(int(year_str))

    if not years:
        print(f"No valid CSV files found for zone: {zone}")
        continue

    ### ### ind the closest year
    closest_year = min(years, key=lambda x: abs(x - data_target_year_int))

    ### ### Construct the filename and path
    closest_csv = f"{closest_year}.csv"
    csv_path = os.path.join(zone_path, closest_csv)

    ### ### Read the CSV into a DataFrame
    df = pd.read_csv(csv_path, header=None)

    ### ### Rename columns
    df.columns = ['snapshot'] + ['load_real_time'] * (len(df.columns) - 1)
    
    ### ### Store in the dictionary
    load_real_time_dict[zone] = df

### Done
print(f"load_real_time_dict :    \n{load_real_time_dict}")
```

<div style="background-color: #000; padding: 20px; border-radius: 8px; border: 1px solid #87CEEB; font-family: 'Times New Roman', Times, serif;">
    <div style="text-align: left; color: #87CEEB;">
        <span style="font-weight: bold; font-size: 20px;: 10px;">
            5.1. Datetime Standardization and Leap Year Normalization
        </span>
    </div>
    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5; border-top: 1px solid rgba(135, 206, 235, 0.3); padding-top: 15px;">
        This routine ensures that all time-series demand data is chronologically consistent with the target simulation year. It performs three critical operations:
        <br>
        • <b>Datetime Standardization:</b> It dynamically detects, cleans, and converts timestamps to UTC, ensuring uniform temporal indexing across all zones.
        <br>
        • <b>Calendar Year Update:</b> It maps existing data to the <code>data_target_year</code>, using safety checks to drop invalid dates (e.g., Feb 29th) when moving to a non-leap year.
        <br>
        • <b>Leap Year Synthesis:</b> If the target year is a leap year, the script intelligently synthesizes a February 29th profile by interpolating values between February 28th and March 1st.
    </div>
        <br>
    <div style="text-align: justify; color: #87CEEB; font-size: 13px; line-height: 1.5;">
        By programmatically managing these calendar discrepancies, the script prevents temporal alignment errors, guaranteeing that the load data is perfectly formatted for the time-steps required by the energy model.
    </div>
</div>

```python
### Define the functions for year updating and to manage the leap year situation
def is_leap_year(year):
    return calendar.isleap(year)
def standardize_and_update_datetime(df, target_year):
    
    ### ### Identify the datetime column
    datetime_col = None
    for col in df.columns:
        if pd.api.types.is_datetime64_any_dtype(df[col]):
            datetime_col = col
            break
            
        ### ### ### Updated regex to also catch shorter formats like '2030-01-01 00'
        elif df[col].dtype == 'object' and df[col].astype(str).str.contains(r'\d{4}-\d{2}-\d{2}', na=False).any():
            datetime_col = col
            break
    if datetime_col is None:
        raise ValueError("No datetime column found in the DataFrame.")

    ### ### Standardize datetime format and enforce UTC (+00:00) | Convert to datetime object
    df.loc[:, datetime_col] = pd.to_datetime(df[datetime_col], errors='coerce')

    ### ### If to_datetime fails to change the dtype to datetime64 force it. This prevents the ".dt" accessor error
    if not pd.api.types.is_datetime64_any_dtype(df[datetime_col]):
        df[datetime_col] = pd.to_datetime(df[datetime_col])

    ### ### Only attempt timezone standardization if we actually have valid dates
    if not df[datetime_col].isna().all():
        if df[datetime_col].dt.tz is None:
            
            ### ### ### ### Localize naive timestamps to UTC
            df[datetime_col] = df[datetime_col].dt.tz_localize('UTC')
        else:
            
            ### ### ### ### Convert existing timezones to UTC
            df[datetime_col] = df[datetime_col].dt.tz_convert('UTC')

    ### ### Update the year
    if not df[datetime_col].isna().all():
        def update_year(x):
            if pd.isna(x): return None
            year, month, day = x.year, x.month, x.day
            
            ### ### ### ### Check if the day exists in the target year (prevents Feb 29 crash on non-leap years)
            if day <= calendar.monthrange(int(target_year), month)[1]:
                return x.replace(year=int(target_year))
            else:
                return None
        df.loc[:, datetime_col] = df[datetime_col].apply(update_year)
        df = df.dropna(subset=[datetime_col]).copy()

    ### ### Add Feb 29 if target year is a leap year
    if is_leap_year(int(target_year)):
        feb29_exists = ((df[datetime_col].dt.month == 2) & (df[datetime_col].dt.day == 29)).any()
        if not feb29_exists:
            
            ### ### ### ### Create Feb 29 timestamps for each hour
            feb29_timestamps = pd.date_range(
                f"{target_year}-02-29 00:00:00+00:00",
                f"{target_year}-02-29 23:00:00+00:00",
                freq="h"
            )

            ### ### ### ###  Get Feb 28 and Mar 1 data for interpolation
            feb28_data = df[(df[datetime_col].dt.month == 2) & (df[datetime_col].dt.day == 28)]
            mar1_data = df[(df[datetime_col].dt.month == 3) & (df[datetime_col].dt.day == 1)]

            ### ### ### ### Safety check: ensure we have 24 hours of data to interpolate from
            if len(feb28_data) >= 24 and len(mar1_data) >= 24:
                new_rows = []
                for i in range(24):
                    feb28_row = feb28_data.iloc[i]
                    mar1_row = mar1_data.iloc[i]

                    ### ### ### ### ### ### Create new rows for Feb 29
                    new_row = {}
                    for col in df.columns:
                        if col == datetime_col:
                            new_row[col] = feb29_timestamps[i]
                        elif pd.api.types.is_numeric_dtype(df[col]):
                            new_row[col] = (feb28_row[col] + mar1_row[col]) / 2
                        else:
                            new_row[col] = feb28_row[col]  # Copy non-numeric columns as-is
                    new_rows.append(new_row)

                ### ### ### ### ### Append new rows
                new_df = pd.DataFrame(new_rows)
                df = pd.concat([df, new_df], ignore_index=True)
                df = df.sort_values(by=datetime_col).reset_index(drop=True)
    return df

### Apply to each DataFrame
for zone in zone_names:
    if zone in load_real_time_dict:
        load_real_time_dict[zone] = standardize_and_update_datetime(
            load_real_time_dict[zone], data_target_year
        )
        
### Done 
print("Year update and leap year normalization complete.")
print(f"load_real_time_dict : \n {load_real_time_dict}")
```

            7. Extraction and Normalization of Link Consumption Profiles
        </span>
    </div>
    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5; border-top: 1px solid rgba(135, 206, 235, 0.3); padding-top: 15px;">
        This block processes the master <code>link_consumption</code>—electricity consumption time series derived from the hourly PyPSA results for each unit and link—CSV to create a structured dictionary of zone-specific DataFrames. The process involves:
        <br>
        • <b>Flexible Column Matching:</b> It intelligently scans the raw CSV for headers containing either the full zone name or its corresponding acronym (sourced from <code>zone_names_equivalences_dict</code>), ensuring robust data retrieval even if source labeling conventions vary.
        <br>
        • <b>Data Normalization:</b> Timestamps are standardized into a consistent <code>YYYY-MM-DD HH:MM:SS</code> string format, providing a clean baseline for temporal operations.
        <br>
        • <b>Header Sanitization:</b> The script programmatically strips the zone/acronym identifiers from column names, resulting in standardized headers that facilitate consistent processing across the entire model pipeline.
        <br>
        <br>
        <div style="text-align: justify; color: #87CEEB; font-size: 13px; line-height: 1.5;">
            By isolating this extraction logic, it ensures that the complex master link data is correctly partitioned and cleaned, making it ready for integration with the primary load profiles.
        </div>
    </div>
</div>

```python
### Identify the CSV file
csv_file = None
for file in os.listdir(load_real_time_pypsa_raw_data_folder_path):
    if 'link_consumption' in file and str(data_target_year) in file and file.endswith('.csv'):
        csv_file = file
        break
if not csv_file:
    raise FileNotFoundError(f"No matching CSV found in {load_real_time_pypsa_raw_data_folder_path}")
csv_path = os.path.join(load_real_time_pypsa_raw_data_folder_path, csv_file)

### Load and Format Timestamp
df_raw = pd.read_csv(csv_path)
first_col_name = df_raw.columns[0]
df_raw[first_col_name] = pd.to_datetime(df_raw[first_col_name]).dt.strftime('%Y-%m-%d %H:%M:%S')

### Process Zones and Clean Headers
link_consumption_dict = {}
for zone in zone_names:
    
    ### ### Retrieve potential Acronym
    acronym_list = zone_names_equivalences_dict.get(zone, {}).get("Acronym", [])
    acronym = acronym_list[0].strip() if acronym_list and acronym_list[0].strip() else None
    
    ### ### Identify matching columns
    matched_cols = [col for col in df_raw.columns if zone in col]
    string_to_erase = zone # Default string to remove
    
    ### ### Fallback to Acronym if no columns found for Zone
    if not matched_cols and acronym:
        matched_cols = [col for col in df_raw.columns if acronym in col]
        string_to_erase = acronym # Update string to remove
    if matched_cols:
        
        ### ### ### Create the DataFrame
        zone_df = df_raw[[first_col_name] + matched_cols].copy()
        
        ### ### ### Clean headers
        zone_df.columns = [col.replace(string_to_erase, "").strip("_ ") if col != first_col_name else col for col in zone_df.columns]
        
        ### ### ### Store in dictionary
        link_consumption_dict[zone] = zone_df

### Done
print(f"Dictionary 'link_consumption_dict' ready. Headers have been cleaned of Zone/Acronym identifiers.")
print(f"link_consumption_dict : \n {link_consumption_dict}")
```

        This block iterates through the <code>link_consumption_dict</code> to apply the <code>standardize_and_update_datetime</code> routine to each zone's specific consumption profile. The process serves as the final temporal calibration step:
        <br>
        • <b>Chronological Synchronization:</b> It forces all zone-specific consumption DataFrames to adhere to the target year structure, resolving any discrepancies arising from the original source files' dates.
        <br>
        • <b>Structural Consistency:</b> By standardizing these files alongside the primary load data, the script ensures that all model inputs share a common temporal index.
        <br>
        • <b>Error Prevention:</b> It validates the integrity of the datetime series—handling leap year shifts and timezone localization—ensuring the Dispa-SET engine receives time-series data that is both syntactically correct and chronologically accurate.
        <br>
        <br>
        <div style="text-align: justify; color: #87CEEB; font-size: 13px; line-height: 1.5;">
            By unifying the temporal attributes of the link consumption profiles, this script prepares the datasets for seamless merging with the main electricity load profiles, ensuring internal consistency across the model entire time horizon.
        </div>
    </div>
</div>

```python
### Apply the function for standardizing datetime formats and updating the year to each DataFrame in the dictionary
for zone in zone_names:
    if zone in link_consumption_dict:
        link_consumption_dict[zone] = standardize_and_update_datetime(
            link_consumption_dict[zone], data_target_year
        )

### Done 
print(f"link_consumption_dict : \n {link_consumption_dict}")
```

<div style="background-color: #000; padding: 20px; border-radius: 8px; border: 1px solid #87CEEB; font-family: 'Times New Roman', Times, serif;">
    <div style="text-align: left; color: #87CEEB;">
        <span style="font-weight: bold; font-size: 20px;">
          8. Aggregation of Technology Consumption Profiles
        </span>
    </div>
    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5; border-top: 1px solid rgba(135, 206, 235, 0.3); padding-top: 15px;">
        This block processes the cleaned zone-specific DataFrames to synthesize a total electric consumption profile for each zone. The logic performs the following:
        <br>
        • <b>Technology Filtering:</b> Using the <code>electricity_consumers_tech_dict</code> as a lookup, the script identifies all columns corresponding to valid electricity-consuming technologies.
        <br>
        • <b>Vectorized Summation:</b> It calculates the total demand (<code>Total_Tech_Consumption</code>) by summing across all identified technology columns for every hourly timestamp (<code>axis=1</code>).
        <br>
        • <b>Dictionary Mapping:</b> The resulting profiles, containing only the temporal index and the aggregated consumption value, are stored in <code>aggregated_link_consumption_dict</code> for streamlined access.
        <br>
        <br>
        <div style="text-align: justify; color: #87CEEB; font-size: 13px; line-height: 1.5;">
            By reducing multi-technology consumption data into a singular "Total" hourly value per zone, this process simplifies the final input requirements for the Dispa-SET engine, ensuring that all technology-specific demands are correctly accounted for in the system net load.
        </div>
    </div>
</div>

```python
### Initialize the final dictionary for aggregated data
aggregated_link_consumption_dict = {}
for zone in zone_names:
    
    ### ### Access the specific DataFrame for the current zone
    if zone not in link_consumption_dict:
        print(f"Warning: Zone {zone} not found in link_consumption_dict. Skipping.")
        continue
    df_zone = link_consumption_dict[zone]
    
    ### ### Identify the first column (the formatted timestamp)
    timestamp_col = df_zone.columns[0]
    
    ### ### Find all columns that contain any tech from 'electricity_consumers_tech_dict'
    tech_columns = [
        col for col in df_zone.columns 
        if any(tech in col for tech in electricity_consumers_tech_dict) and col != timestamp_col
    ]
    if tech_columns:
        
        ### ### ### Create the new DataFrame
        new_df = pd.DataFrame()
        new_df[timestamp_col] = df_zone[timestamp_col]
        new_df['Total_Tech_Consumption'] = df_zone[tech_columns].sum(axis=1)
        
        ### ### ### Save the result back into the dictionary
        aggregated_link_consumption_dict[zone] = new_df
    else:
        print(f"Note: No matching technology columns found for zone {zone}.")

### Done
print(f"aggregated_link_consumption_dict : \n {aggregated_link_consumption_dict}")
```

    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5;">

12.2.  Seasonal Heat Demand Disaggregation (DHW vs. SH)
        </span>
    </div>
    <div style="text-align: justify; color: #87CEEB; font-size: 15px; line-height: 1.5; border-top: 1px solid rgba(135, 206, 235, 0.3); padding-top: 15px;">
        This routine separates the total thermal consumption into Domestic Hot Water (DHW) and Space Heating (SH) profiles. The logic follows a thermal load-decomposition heuristic:
        <br>
        • <b>DHW Baseline Extraction:</b> The script assumes that the average demand during summer months (June–August) represents the baseline DHW requirement, which is then shaped by a daily average profile and smoothed over a 24-hour window to remove anomalies.
        <br>
        • <b>Energy Conservation:</b> It performs a normalization step to ensure the total integrated energy of the derived DHW series matches the original baseline, preserving total caloric demand.
        <br>
        • <b>Residual Space Heating:</b> SH demand is calculated as the residual (<code>Total - DHW</code>), with a floor set at zero to ensure no negative demand occurs due to approximation variances.
        <br>
        <br>
        <div style="text-align: justify; color: skyblue; font-size: 13px; line-height: 1.5;">
            By distinguishing between these two demand types, the Dispa-SET engine can apply specific flexibility strategies—such as hot water tank storage for DHW or building thermal inertia modeling for SH—significantly increasing the model's physical fidelity.
        </div>
    </div>
</div>

```python
if vpp == "Y":

    ### ### Seasonal Parameters    
    SUMMER_MONTHS = [6, 7, 8]
    SMOOTHING_HOURS = 24
    
    ### ### Seasonal Spleating Function   
    def split_heat_demand_seasonal(heat_series, summer_months=SUMMER_MONTHS, smoothing_window=SMOOTHING_HOURS):
        heat = heat_series.copy()
        heat.index = pd.to_datetime(heat.index)

        ### ### ### Summer baseline = DHW
        summer_mask = heat.index.month.isin(summer_months)
        dhw_baseline = heat[summer_mask].mean()
        dhw = pd.Series(dhw_baseline, index=heat.index)

        ### ### ### Daily profile
        hourly_profile = heat.groupby(heat.index.hour).mean() / heat.groupby(heat.index.hour).mean().mean()
        dhw *= heat.index.hour.map(hourly_profile)

        ### ### ### Smooth DHW
        dhw = dhw.rolling(smoothing_window, center=True, min_periods=1).mean()

        ### ### ### Normalize energy after smoothing
        target_energy = dhw_baseline * len(dhw)
        dhw *= target_energy / dhw.sum()

        ### ### ### Space heating = residual
        sh = heat - dhw
        sh[sh < 0] = 0

        ### ### ### Final correction
        dhw = heat - sh
        return dhw, sh

    ### ### Calling Seasonal Spleating Function   
    for zone in zone_names:
        df = aggregated_link_heating_consumption_dict[zone]
        
        ### ### ### Ensure timestamp is datetime and set as index
        df["snapshot"] = pd.to_datetime(df["snapshot"])
        df = df.set_index("snapshot")
        
        ### ### ### Extract heat series
        heat_series = df["Total_heating_Consumption"]
        
        ### ### ### Apply your split function
        dhw, sh = split_heat_demand_seasonal(heat_series)
        
        ### ### ### Store results in dataframe
        df["DHW_Consumption"] = dhw
        df["SH_Consumption"] = sh
        
        ### ### ###  (Optional) reset index back if needed
        df = df.reset_index()
        
        ### ### ### Save back to dictionary
        aggregated_link_heating_consumption_dict[zone] = df
        
### Done
print(f"aggregated_link_heating_consumption_dict : \n {aggregated_link_heating_consumption_dict}")
```

```python
if vpp == "Y":

    ### ### Define helper functions
    def find_matching_columns(columns, keyword):
        pattern = re.compile(re.escape(keyword), re.IGNORECASE)
        return [c for c in columns if pattern.search(c)]
    
    def build_clean_name(col1, col2):
        parts1 = col1.split("_")
        parts2 = col2.split("_")
        merged = parts1[:]
        for p in parts2:
            if p not in merged:
                merged.append(p)
        return "_".join(merged)
    
    def ensure_numeric(df, col):
        df[col] = pd.to_numeric(df[col], errors="coerce")
        return df[col]

    ### ### Main Loop   
    for zone in zone_names:
        df = aggregated_link_heating_consumption_dict[zone].copy()
    
        ### ### ### Ensure datetime safety
        if "snapshot" in df.columns:
            df["snapshot"] = pd.to_datetime(df["snapshot"], errors="coerce")
        cols = df.columns
        numeric_cols = df.select_dtypes(include="number").columns
        new_columns = {}
    
        ### ### ### Loop over technology keys (Fractions)
        for tech_key, subdict in heat_suppliers_tech_master_dict.items():
            fraction_cols = [
                c for c in find_matching_columns(cols, tech_key)
                if "fraction" in c and c in numeric_cols
            ]
            if not fraction_cols:
                continue

            ### ### ### ### Lopp over Alpha Sub-Keys
            for alpha_key, alpha_value in subdict.items():
                if not alpha_key.startswith("alpha_"):
                    continue
                alpha_suffix = alpha_key.replace("alpha_", "")
                alpha_cols = [
                    c for c in find_matching_columns(cols, alpha_suffix)
                    if "fraction" not in c and c in numeric_cols
                ]
                for fcol in fraction_cols:
                    for acol in alpha_cols:
                        ensure_numeric(df, fcol)
                        ensure_numeric(df, acol)
                        new_col_name = build_clean_name(fcol, acol)
                        new_columns[new_col_name] = df[fcol] * df[acol]

        ### ### ### Add New Columns        
        for k, v in new_columns.items():
            df[k] = v
    
        ### ### ### Second Stage    
        generated_cols = list(new_columns.keys())
        second_stage_columns = {}
        for tech_key, subdict in heat_suppliers_tech_master_dict.items():
    
            ### ### ### ### find generated columns containing this tech_key
            matched_generated = find_matching_columns(generated_cols, tech_key)
            if not matched_generated:
                continue
            for col in matched_generated:
                for alpha_key, alpha_value in subdict.items():
                    if not alpha_key.startswith("alpha_"):
                        continue
                    alpha_suffix = alpha_key.replace("alpha_", "")
    
                    ### ### ### ### ### ### apply only if suffix is present in column name
                    if re.search(re.escape(alpha_suffix), col, re.IGNORECASE):
                        new_col_name = f"{col}_weighted"
                        second_stage_columns[new_col_name] = df[col] * alpha_value
    
        ### ### ### Add second-stage columns
        for k, v in second_stage_columns.items():
            df[k] = v
            
        ### ### ### Save back
        aggregated_link_heating_consumption_dict[zone] = df

### Done
print(f"aggregated_link_heating_consumption_dict : \n {aggregated_link_heating_consumption_dict}")
```

        • <b>Conditional Source Selection:</b> Dynamically switches between aggregated technology profiles (if <code>vpp == "Y"</code>) and standard consumption datasets, applying specialized summation logic for each.

        • <b>Temporal Alignment:</b> Identifies and harmonizes datetime columns across disparate source datasets to ensure time-series integrity.
        <br>
        • <b>Year-Proximity Logic:</b> Implements a selection mechanism that defaults to the dataset whose temporal reference year is closest to the defined <code>data_target_year_int</code>, mitigating errors when merging time series from different vintages.
        <br>
        • <b>Synthesis:</b> Merges the processed heating demand with existing demand DataFrames, resulting in a unified structure indexed by the most appropriate temporal reference.

```python
### Define the source of the data based on vpp condition
if vpp == "Y":
    source_dict = aggregated_link_heating_consumption_dict
    
    ### ### Define a helper function to get the specific columns
    get_sum_series = lambda df: df[[c for c in df.columns if c.endswith('SH_weighted')]].sum(axis=1)
else:
    source_dict = aggregated_link_consumption_dict
    
    ### ### Get the only non-date column
    get_sum_series = lambda df: df[[c for c in df.columns if 'snap' not in c.lower() and 'date' not in c.lower() and 'time' not in c.lower()]].iloc[:, 0]

### Single loop to process all zones
for zone in zone_names:
    demand_main_key = list(demand_data_with_dataframes_dict.keys())[0]
    paths_main_key = list(demand_data_paths_dict.keys())[0]
    
    ### ### Get DataFrames
    df_src = source_dict[zone].copy()
    
    ### ### Find the demand DataFrame in the nested dict
    zone_data_struct = demand_data_with_dataframes_dict[demand_main_key][zone]
    target_ssk = next(k for k, v in zone_data_struct.items() if isinstance(v, pd.DataFrame))
    df_demand = zone_data_struct[target_ssk].copy()

    ### ### Identify date columns
    date_col_src = [c for c in df_src.columns if any(x in c.lower() for x in ['snap', 'date', 'time'])][0]
    date_col_dmd = [c for c in df_demand.columns if any(x in c.lower() for x in ['snap', 'date', 'time'])][0]
    df_src[date_col_src] = pd.to_datetime(df_src[date_col_src])
    df_demand[date_col_dmd] = pd.to_datetime(df_demand[date_col_dmd])

    ### ### Perform the calculation (Dynamic summation)
    summed_values = get_sum_series(df_src) + df_demand.select_dtypes(include=['number']).iloc[:, 0]

    #### ### ear Proximity logic
    year_src = df_src[date_col_src].dt.year.iloc[0]
    year_dmd = df_demand[date_col_dmd].dt.year.iloc[0]
    final_dates = df_src[date_col_src] if abs(year_src - data_target_year_int) <= abs(year_dmd - data_target_year_int) else df_demand[date_col_dmd]

    ### ### Store result
    demand_data_paths_dict[paths_main_key][zone][target_ssk] = pd.DataFrame({
        date_col_dmd: final_dates,
        paths_main_key: summed_values
    })

### Done
print(f"Update complete. demand_data_paths_dict modified.")
print(f"demand_data_paths_dict : \n {demand_data_paths_dict}")
```

        <br><br>
        • <b>Categorical Summation:</b> Iteratively scans the processed dataframes and performs a column-wise sum based on the <code>SH_weighted</code> and <code>DHW_weighted</code> suffixes.
        <br>
        • <b>Dictionary Structuring:</b> Builds the <code>dhw_sh_consumption_dict</code>, providing a clean, nested interface where each <code>zone</code> maps to discrete <code>SH</code> and <code>DHW</code> DataFrames.

```python
### Initialize the single dictionary
dhw_sh_consumption_dict = {}

if vpp == "Y":

    ### ### Iterate through the heating dictionary
    for zone, df in aggregated_link_heating_consumption_dict.items():
        
        ### ### ### Identify the datetime column
        date_col = df.columns[df.columns.str.contains('snap|date|time', case=False)][0]
        
        ### ### ### Extract and sum column groups
        sh_sum = df[[c for c in df.columns if c.endswith('SH_weighted')]].sum(axis=1)
        dhw_sum = df[[c for c in df.columns if c.endswith('DHW_weighted')]].sum(axis=1)
        
        ### ### ### Initialize the zone key in the new dictionary
        dhw_sh_consumption_dict[zone] = {}
        
        ### ### ### Store the results in sub-keys 'SH' and 'DHW'
        dhw_sh_consumption_dict[zone]['SH'] = pd.DataFrame({
            date_col: df[date_col],
            'Total_SH_Consumption': sh_sum
        })
        dhw_sh_consumption_dict[zone]['DHW'] = pd.DataFrame({
            date_col: df[date_col],
            'Total_DHW_Consumption': dhw_sum
        })
    
### Done
print(f"dhw_sh_consumption_dict successfully built for zones: {list(dhw_sh_consumption_dict.keys())}")
print(f"DHW Dictionary created with keys :  \n {dhw_sh_consumption_dict}")
```
